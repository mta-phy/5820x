```python
from matplotlib import pyplot as plt
import pandas as pd

data=pd.read_csv('element_properties.csv')


data1=data['ionization energy'][1:22]
data2=data['ionization energy'][22:]
plt.boxplot([data1,data2],positions=[1,2])
plt.xticks([1,2],['metal','nonmetal'])
plt.xlabel("Group")
plt.ylabel('Ionization energy')
plt.show()
```

```python
data1=data['atomic_radius'][0:21]
data2=data['atomic_radius'][21:]
plt.boxplot([data1,data2],positions=[1,2])
plt.xticks([1,2],['metal','nonmetal'])
plt.xlabel("Group")
plt.ylabel('Atomic Radius')
plt.show()
```
