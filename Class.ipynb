```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Parameters
# -----------------------------
a = 1.0
t = 1.0

# -----------------------------
# k-space grid
# -----------------------------
k = np.linspace(-np.pi, np.pi, 300)
kx, ky = np.meshgrid(k, k)

# -----------------------------
# Dispersion
# -----------------------------
E = -2 * t * (np.cos(kx * a) + np.cos(ky * a))

# -----------------------------
# (a) 3D Plot with colormap
# -----------------------------
fig = plt.figure(figsize=(8, 14))
ax = fig.add_subplot(projection='3d')

surf = ax.plot_surface(kx, ky, E, cmap='viridis')

ax.set_xlabel('kx')
ax.set_ylabel('ky')
ax.set_zlabel('E(k) in units of t')
ax.set_title('E(k) vs. k')

# Add colorbar
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

plt.show()

# -----------------------------
# (b) Diagonal band
# -----------------------------
k_diag = np.linspace(-np.pi, np.pi, 300)
E_diag = -4 * t * np.cos(k_diag * a)

fig=plt.figure(figsize=(8, 6))
plt.plot(k_diag, E_diag, color='blue')
plt.xlabel('k (kx = ky)')
plt.ylabel('E(k) in units of t')
plt.title('Band along Zone Diagonal')
plt.grid()
plt.show()

# -----------------------------
# (c) Fermi surfaces with colors
# -----------------------------
plt.figure()

E_F_values = [0, 2, -2]
colors = ['red', 'green', 'blue']

for EF, color in zip(E_F_values, colors):
    CS = plt.contour(kx, ky, E, levels=[EF], colors=color)
    CS.collections[0].set_label(f'E_F = {EF}')

plt.xlabel('kx')
plt.ylabel('ky')
plt.title('Fermi Surfaces (Colored)')
plt.legend()
plt.gca().set_aspect('equal')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters based on Problem 1
epsilon_0 = 0  # Center of the band
t = 1          # Set t=1
a = 1          # Lattice constant

# Create k grid in units of pi/a
k_val = np.linspace(-np.pi, np.pi, 500)
kx, ky = np.meshgrid(k_val, k_val)

# Tight-binding dispersion formula
epsilon = -epsilon_0 - 2*t*(np.cos(kx) + np.cos(ky))

# Define the specific Fermi energies for 1c[cite: 1]
fermi_levels = [
    -epsilon_0 - 2*t,  # Case: -epsilon_0 - 2t[cite: 1]
    -epsilon_0,        # Case: -epsilon_0[cite: 1]
    -epsilon_0 + 2*t   # Case: -epsilon_0 + 2t[cite: 1]
]

titles = [
    r'Fermi Surface: $\epsilon_F = -2t$', 
    r'Fermi Surface: $\epsilon_F = 0$', 
    r'Fermi Surface: $\epsilon_F = +2t$'
]

colors = ['blue', 'black', 'red']

# Create the figure with 3 subplots for 1c
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i in range(3):
    # Plot the specific contour for the Fermi level[cite: 1]
    axes[i].contour(kx/np.pi, ky/np.pi, epsilon, levels=[fermi_levels[i]], colors=colors[i])
    
    # Formatting each subplot
    axes[i].set_title(titles[i])
    axes[i].set_xlabel(r'$k_x$ ($\pi/a$)')
    axes[i].set_ylabel(r'$k_y$ ($\pi/a$)')
    axes[i].set_aspect('equal')
    axes[i].axhline(0, color='gray', lw=0.5, ls='--')
    axes[i].axvline(0, color='gray', lw=0.5, ls='--')
    axes[i].grid(True, linestyle=':', alpha=0.6)

plt.suptitle('Problem 1c: Fermi Surfaces at Specified Constant Energies', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
