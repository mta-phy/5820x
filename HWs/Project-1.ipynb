```python
import pandas as pd
import numpy as np
url = 'https://raw.githubusercontent.com/mta-phy/5820x/main/MDS-data/mdsdata/DS1_Iris/iris.csv'
Data = pd.read_csv(url, header=None)
S_length = Data.iloc[:, 0].to_numpy();
S_width = Data.iloc[:, 1].to_numpy();
P_length = Data.iloc[:, 2].to_numpy();
P_width = Data.iloc[:, 3].to_numpy();
Class=Data.iloc[:, 4].to_numpy();
target_map = {'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2}
target = Data.iloc[:, 4].map(target_map).to_numpy()

df = pd.DataFrame({
    'Sepal Length (cm)': S_length,
    'Sepal Width (cm)': S_width,
    'Petal Length (cm)': P_length,
    'Petal Width (cm)': P_width,
    'Class':Class})
df[:4]
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
from scipy.stats import norm

def gaussian(x, loc, s):
    return (1 / np.sqrt(2 * np.pi * s ** 2) * np.exp(-(x - loc) ** 2 / (2 * s ** 2)))

plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.hist(S_length, bins=10, density=True, alpha=0.7, color='C0', edgecolor='black')
x0_range = np.linspace(S_length.min(), S_length.max(), 100)
y0_values = gaussian(x0_range, S_length.mean(), S_length.std())
plt.plot(x0_range, y0_values, 'r-', linewidth=2, label='Custom Gaussian Fit')
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Probability density")
plt.title("Sepal Length distribution")

plt.subplot(2, 2, 2)
plt.hist(S_width, bins=10, density=True, alpha=0.7, color='C1', edgecolor='black')
x1_range = np.linspace(S_width.min(), S_width.max(), 100)
y1_values = gaussian(x1_range, S_width.mean(), S_width.std())
plt.plot(x1_range, y1_values, 'r-', linewidth=2, label='Custom Gaussian Fit')
plt.xlabel("Sepal Width (cm)")
plt.ylabel("Probability density")
plt.title("Sepal Width distribution")

plt.subplot(2, 2, 3)
plt.hist(P_length, bins=10, density=True, alpha=0.7, color='C2', edgecolor='black')
x2_range = np.linspace(P_length.min(), P_length.max(), 100)
y2_values = gaussian(x2_range, P_length.mean(), P_length.std())
plt.plot(x2_range, y2_values, 'r-', linewidth=2, label='Custom Gaussian Fit')
plt.xlabel("Petal Length (cm)")
plt.ylabel("Probability density")
plt.title("Petal Length distribution")

plt.subplot(2, 2, 4)
plt.hist(P_width, bins=10, density=True, alpha=0.7, color='C3', edgecolor='black')
x3_range = np.linspace(P_width.min(), P_width.max(), 100)
y3_values = gaussian(x3_range, P_width.mean(), P_width.std())
plt.plot(x3_range, y3_values, 'r-', linewidth=2, label='Custom Gaussian Fit')
plt.xlabel("Petal Width (cm)")
plt.ylabel("Probability density")
plt.title("Petal Width distribution")

plt.show()
```

Corelation Matrix

```python
cor = np.corrcoef(df.iloc[:, 0:4], rowvar=False)

# Print correlation matrix
print("Correlation Matrix:\n", cor)
```

```python
plt.figure(figsize=(6,6))
feature_names=['Sepal Length','Sepal width','Petal Length','Petal width']
im = plt.imshow(cor, vmin=-0.45, vmax=1, cmap='terrain')
plt.colorbar(im, label="Correlation coefficient")
plt.xticks(range(len(feature_names)), feature_names, rotation=45)
plt.yticks(range(len(feature_names)), feature_names)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()

```
