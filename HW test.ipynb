# Chap 7.2 (b)

```python
import numpy as np
import math
import matplotlib.pyplot as plt

k_B = 1.3806e-23  # in J/K
T = [300, 350, 400, 450, 500, 550, 600]           # in Kelvin
amu_to_kg = 1.6605e-27
m = 4 * amu_to_kg  # Mass of helium atom in kg

# Maxwell-Boltzmann speed distribution function
def f_MB(c, m, T):
    factor = 4 * np.pi * c**2 * (m / (2 * np.pi * k_B * T))**(3/2)
    exponent = np.exp(-m * c**2 / (2 * k_B * T))
    return factor * exponent

T_array = np.array(T).reshape(-1, 1)

c = np.linspace(0, 3000, 1000)
f_c = f_MB(c, m, T_array)

dc=c[1]-c[0]
Area=np.sum(f_c*dc, axis=1)
Mean=np.sum(c*f_c*dc, axis=1)/Area
Var=np.sum(dc * f_c * (c - Mean[:, None])**2, axis=1)
SD=np.sqrt(Var)

plt.bar(T, SD, label='He', width=25)

plt.xlabel('Temperature (K)')
plt.ylabel('Standard Deviation (m/s)')
plt.title('Standard Deviation of Speeds')
plt.legend()
plt.ylim(400, 650)
plt.tight_layout()
plt.show()
```
