```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris=load_iris()
X, y=iris.data, iris.target
```

```python
X_train, X_test, y_train, y_test=train_test_split(X, y, test_size=0.2)
```

```python
model=KNeighborsClassifier(n_neighbors=4)
model.fit(X_train,y_train)
```

```python
y_pred=model.predict([[5,4,3,1]])
y_pred
```

```python
print(f"Accuracy: {accuracy_score(y_test,y_pred)}")
```

```python
import matplotlib.pyplot as plt

# Feature indices:
# 0 = sepal length
# 1 = sepal width

plt.figure(figsize=(8,6))

for species in [0, 1, 2]:
    plt.scatter(
        X[y == species, 0],   # sepal length
        X[y == species, 1],   # sepal width
        label=iris.target_names[species]
    )
plt.scatter(5,4, marker='d')

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Sepal Width vs Sepal Length (by Species)")
plt.legend()
plt.grid(True)
plt.show()
```
