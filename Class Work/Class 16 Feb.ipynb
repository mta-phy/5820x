```python
import pandas as pd
import numpy as np

data =pd.read_excel("LIG_resistances.xlsx", header=None)
population=data.iloc[:,0].to_numpy()
print("Mean: ", population.mean())
print(population[:10])
```

```python
#Simple Random number

rng=np.random.default_rng(42)
sample=rng.choice(population, size=15, replace=False)
print("Sample: ", sample)
print("Mean: ", np.mean(sample))
```

```python
#Multiple Independent Samples
Size=5

rng=np.random.default_rng(42)
s1=rng.choice(population, size=Size, replace=False)
s2=rng.choice(population, size=Size, replace=False)
s3=rng.choice(population, size=Size, replace=False)
print("S1: ", s1)
print("S2: ", s2)
print("S3: ", s3)
```
