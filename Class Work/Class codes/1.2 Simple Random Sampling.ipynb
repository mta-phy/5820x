```python
# Cell 1: Import libraries
import numpy as np

# Create a random number generator
rng = np.random.default_rng(42)

# Simulate a population of LIG conductivities (S/cm)
#X~N(mean, sigma^2)
# Assume 2000 LIG samples, mean ~180 S/cm, sigma ~10
population = rng.normal(loc=180, scale=10, size=2000)

# Compute the true population mean (for reference)
true_mean = population.mean()
print(f"True population mean conductivity: {true_mean:.2f} S/cm")
```

```python
# Cell 2: Simple random sample of 20 LIG measurements
sample_size = 20
sample = rng.choice(population, size=sample_size, replace=False)

# Compute sample mean
sample_mean = sample.mean()
print(f"Sample values: {np.round(sample,1)}")
print(f"Sample mean conductivity: {sample_mean:.2f} S/cm")
```

```python
# Cell 3: Generate multiple independent random samples
n_samples = 25  # number of samples to draw
sample_size = 20

sample_means = []

for i in range(n_samples):
    s = rng.choice(population, size=sample_size, replace=False)
    sample_means.append(s.mean())

print(f"Sample means from {n_samples} independent samples: {np.round(sample_means,2)}")
```

```python
# Cell 4: Histogram to visualize population and samples
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.hist(population, bins=30, alpha=0.5, label='Population', color='skyblue')
plt.hist(sample, bins=10, alpha=0.8, label='Single Sample', color='orange')
plt.axvline(true_mean, color='blue', linestyle='--', label='True mean')
plt.axvline(sample_mean, color='red', linestyle='--', label='Sample mean')
plt.xlabel("LIG Conductivity (S/cm)")
plt.ylabel("Frequency")
plt.title("Population vs Simple Random Sample")
plt.legend()
plt.show()
```
