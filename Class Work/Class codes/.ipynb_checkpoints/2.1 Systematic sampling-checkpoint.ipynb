```python
import numpy as np
import pandas as pd
import math
```

```python
# Load Excel file
data = pd.read_excel("LIG_resistances.xlsx", header=None)

# Convert to 1D NumPy array
population = data.values.flatten()

# Remove empty cells if any
population = population[~pd.isna(population)]

# Convert to float
population = population.astype(float)
N = len(population)

# Check first few values
print("First 10 values of population:", population[:10])
print("Population size (N):", len(population))
print(N)
```

```python
# Desired sample size
n = 6  # change if needed

# Sampling interval k = ceil(N / n)
k = math.ceil(N / n)

print("Sampling interval (k):", k)
```

```python
# Random number generator
rng = np.random.default_rng()

# Random start index
start = rng.integers(0, N)

# Extend population to allow wrap-around
extended_population = np.tile(population, 2)

print("Random start index:", start)
```

```python
# Pick every k-th element starting from start, take first n
systematic_sample = extended_population[start::k][:n]

print("Systematic Sample:", systematic_sample)
```

```python
systematic_mean = systematic_sample.mean()
print("Systematic Sample Mean:", systematic_mean)
```

```python
# Simple Random Sampling for comparison
srs_sample = rng.choice(population, size=n, replace=False)

print("SRS Sample:", srs_sample)
print("SRS Mean:", srs_sample.mean())
```
