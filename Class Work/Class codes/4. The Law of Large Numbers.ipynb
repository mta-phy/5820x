```python
#The Law of Large Numbers: Example with LIG Conductivity Measurements. 
#As a realistic experimental example of the law of large numbers, consider repeated measurements of the electrical conductivity of 
#laser-induced graphene (LIG) fabricated under identical laser parameters. Due to instrumental noise, local inhomogeneities, and 
#contact resistance variations, individual conductivity measurements fluctuate even when taken on nominally identical samples.
#Suppose we repeatedly measure the conductivity (in S/cm) of LIG produced at a fixed laser power and scan speed.
```

```python
#Creating the Dataset
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
```

```python
#We generate a sequence of conductivity measurements centered around a physically reasonable expected value of approximately 180 S/cm, 
#with realistic experimental noise.

n_samples = 5500
# Simulated repeated conductivity measurements (S/cm)

values = rng.normal(loc=180.0, scale=10.0, size=n_samples)
```

```python
#Here, each value corresponds to one independent conductivity measurement performed under the same experimental conditions.
```

```python
#Cumulative Averages
#We now compute the cumulative mean conductivity after each additional measurement:
sample = np.arange(1, n_samples + 1)
averages = values.cumsum() / sample
```

```python
#The expected (true) mean conductivity is approximately
expected_value = values.mean()
expected_value
```

```python
#Visualization of the Law of Large Numbers
fig, ax = plt.subplots(ncols=2, figsize=(9, 3))

ax[0].plot(sample, values, 'o', mfc='C0', mec='none', alpha=0.4)
ax[0].set_xlabel("Measurement number")
ax[0].set_ylabel("Conductivity (S/cm)")

ax[1].plot(sample, averages - expected_value, '-')
ax[1].set_xlabel("Number of measurements") 
ax[1].set_ylabel("Deviation from mean (S/cm)")
```

```python
#Interpretation

#The left panel shows the strong fluctuations observed in individual conductivity measurements. 
#The right panel displays the deviation of the cumulative mean conductivity from the expected value as a
#function of the number of measurements.
#As predicted by the law of large numbers, the deviation decreases as more measurements are averaged. 
#While early averages may significantly overestimate or underestimate the true conductivity, the cumulative mean
#converges toward the expected value as the sample size increases.
```

```python
#Discussion

#This example demonstrates the fundamental principle underlying experimental reproducibility: averaging many independent measurements reduces 
#random error. If this were not the case, quantitative materials characterization would be impossible.
#At the same time, the law of large numbers does not imply that small datasets provide reliable estimates of the expected value. 
#Early averages may deviate strongly from the true mean, emphasizing that there is no “law of small numbers.”
#In the context of materials science and machine learning, the LLN highlights the importance of representative datasets. 
#If measurements are biased—for example, by systematically probing only low-resistance regions of a LIG film—the cumulative average will converge,
#but to the wrong value.
    
```
