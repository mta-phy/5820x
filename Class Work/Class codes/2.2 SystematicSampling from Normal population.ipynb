```python
# ============================================
# CELL 1: Import required libraries
# ============================================

import numpy as np
import matplotlib.pyplot as plt
```

```python
# ============================================
# CELL 2: Create a normal population
# ============================================

rng = np.random.default_rng(42)  # reproducibility

N = 10000   # population size

# Generate a normal population (mean = 50, std = 10)
population = rng.normal(loc=50, scale=10, size=N)

# Compute true population mean
population_mean = np.mean(population)

print("Population mean:", round(population_mean, 2))
rng
```

```python
# ============================================
# CELL 3: Systematic sampling
# ============================================

n = 20  # sample size

k = N // n   # sampling interval


# Choose random starting point between 0 and k-1
start = rng.integers(0, k)
print(f"K:  {k}")
# Select every k-th element
systematic_sample = population[start::k][:n]

# Compute sample mean
sample_mean = np.mean(systematic_sample)

print("Systematic sample mean:", round(sample_mean, 3))

systematic_sample
```

```python
# ============================================
# CELL 4: Visualization
# ============================================

plt.figure(figsize=(10,6))

# Histogram of full population
plt.hist(population, bins=40, alpha=0.4, density=True, label="Population")

# Histogram of systematic sample
plt.hist(systematic_sample, bins=30, alpha=0.6, density=True, label="Systematic Sample")

# Plot mean lines
plt.axvline(population_mean, linestyle='--', linewidth=2, label="Population Mean")
plt.axvline(sample_mean, linestyle='-', linewidth=2, label="Sample Mean")

plt.title("Systematic Sampling from Normal Population")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.show()
```
