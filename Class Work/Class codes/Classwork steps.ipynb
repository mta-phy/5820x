```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data=pd.read_excel("LIG_realistic_dataset.xlsx")
Fluence = data.iloc[:, 0].to_numpy();
Id_Ig = data.iloc[:, 1].to_numpy();
Resistance = data.iloc[:, 2].to_numpy();
Conductivity = data.iloc[:, 3].to_numpy();
```

```python
plt.hist(Fluence, bins=50, density=True, alpha=0.7, color='C0')
plt.xlabel("Fluence")
plt.ylabel("Probability density")
plt.title("Fluence distribution")
plt.show()
```

```python
plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
plt.scatter(Fluence, Id_Ig, color='C0')
plt.xlabel("Fluence (J/cm²)")
plt.ylabel("ID/IG ratio")
plt.title("Fluence vs ID/IG")

plt.subplot(2,2,2)
plt.scatter(Id_Ig, Resistance, color='C1')
plt.xlabel("ID/IG ratio")
plt.ylabel("Sheet resistance (Ohm/sq)")
plt.title("ID/IG vs Sheet resistance")

plt.subplot(2,2,3)
plt.scatter(Resistance, Conductivity, color='C2')
plt.xlabel("Resistance")
plt.ylabel("Conductivity")
plt.title("R vs G")

plt.tight_layout()
plt.show()
```

```python
cor = np.corrcoef(data, rowvar=False)

# Print correlation matrix
print("Correlation matrix:\n", cor)
```

```python
feature_names = ["Fluence", "ID/IG", "Resistance", "Conductivity"]
plt.figure(figsize=(5,4))
im = plt.imshow(cor, vmin=-1, vmax=1, cmap='viridis')
plt.colorbar(im, label="Correlation coefficient")
plt.xticks(range(len(feature_names)), feature_names, rotation=45)
plt.yticks(range(len(feature_names)), feature_names)
plt.title("Correlation Matrix of LIG Properties")
plt.tight_layout()
plt.show()
```
