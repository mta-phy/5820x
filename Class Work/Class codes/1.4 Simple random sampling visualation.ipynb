```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
LIG sheet resistance population

# Assume 100 LIG electrodes fabricated under identical conditions
# Sheet resistance follows an approximately normal distribution

rng = np.random.default_rng(seed=42)

N = 1000  # population size
true_mean = 35   # ohms/sq (typical LIG value)
true_std = 5     # fabrication variability

lig_population = rng.normal(true_mean, true_std, size=N)

# True population mean (usually unknown in real experiments)
population_mean = lig_population.mean()
population_mean
```

```python

# Population visualization
# ===========================================================
plt.figure(figsize=(6,4))
plt.hist(lig_population, bins=15, edgecolor='k', alpha=0.7)
plt.axvline(population_mean, linestyle='--', linewidth=2, label="Population Mean")

plt.xlabel("Sheet Resistance (Ω/sq)")
plt.ylabel("Count")
plt.title("LIG Sheet Resistance Population")
plt.legend()
plt.show()
```

```python
#Simple Random Sampling (SRS)
# ===========================================================
sample_size = 1000

sample = rng.choice(
    lig_population,
    size=sample_size,
    replace=False  # key property of SRS
)

sample, sample.mean()
```

```python
#Mean comparison
# ===========================================================
print(f"Population mean resistance : {population_mean:.2f} Ω/sq")
print(f"Sample mean resistance     : {sample.mean():.2f} Ω/sq")
```

```python
# Sample vs population visualization
# ===========================================================
plt.figure(figsize=(6,4))
plt.hist(lig_population, bins=15, alpha=0.4, label="Population")
plt.hist(sample, bins=10, alpha=0.9, label="Random Sample")

plt.axvline(population_mean, linestyle='--', linewidth=2, label="Population Mean")
plt.axvline(sample.mean(), linestyle=':', linewidth=2, label="Sample Mean")

plt.xlabel("Sheet Resistance (Ω/sq)")
plt.ylabel("Count")
plt.title("Simple Random Sampling from LIG Population")
plt.legend()
plt.show()
```
