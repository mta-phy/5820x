```python
import numpy as np
import pandas as pd

# Random number generator
rng = np.random.default_rng()

# Step 1: Load Excel file
# Assuming no headers, and each column is a stratum
data = pd.read_excel("LIG_conductivities.xlsx", header=None)

# Convert columns to separate strata
low_power = data.iloc[:, 0].dropna().to_numpy().astype(float) #.dropna() → removes empty cells.
                                                               #.to_numpy().astype(float) → converts column to a NumPy array.
medium_power = data.iloc[:, 1].dropna().to_numpy().astype(float)
high_power = data.iloc[:, 2].dropna().to_numpy().astype(float)

# Step 2: Combine for population
population = np.concatenate([low_power, medium_power, high_power])
print("Population mean:", population.mean())

# Optional: check sizes
print("Low power size:", len(low_power))
print("Medium power size:", len(medium_power))
print("High power size:", len(high_power))
```

```python
#Stratified Sampling
#We now perform stratified sampling by applying simple random sampling within each stratum, drawing samples proportional to the stratum sizes.

n_low = 3
n_medium = 4
n_high = 5

s_low = rng.choice(low_power, size=n_low, replace=False)
s_medium = rng.choice(medium_power, size=n_medium, replace=False)
s_high = rng.choice(high_power, size=n_high, replace=False)

sample = np.concatenate([s_low, s_medium, s_high])
sample
```

```python
#Computing the Population Mean
population.mean()
```
