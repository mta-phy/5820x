```python
import pandas as pd
import numpy as np

# Read Excel file
data = pd.read_excel("LIG_resistances.xlsx")

# Convert first column to numpy array
population = data.iloc[:, 0].to_numpy()

print("Mean:", population.mean())
```

```python
#Simple Random Sample
rng = np.random.default_rng(42) #42 is the seed: it ensures that the random numbers are reproducible.
sample = rng.choice(population, size=15, replace=False) # picks random elements from an array.

sample
```

```python
sample.mean()
```

```python
#Multiple Independent Samples
n = 5

s1 = rng.choice(population, size=n, replace=False)
s2 = rng.choice(population, size=n, replace=False)
s3 = rng.choice(population, size=n, replace=False)

print("s1:", s1)
print("s2:", s2)
print("s3:", s3)
```

```python
print("mean values:",
      s1.sum() / n,
      s2.sum() / n,
      s3.sum() / n)
```
